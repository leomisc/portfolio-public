```python
import pandas as pd
from datetime import date
import numpy as np

df = pd.read_csv("data/raw/Sample - Superstore.csv", encoding="cp1252", dtype={"Postal Code": "str"}, index_col=0)
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y").dt.normalize()
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y").dt.normalize()
```
