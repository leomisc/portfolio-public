```python
import pandas as pd
from datetime import date
import numpy as np

df = pd.read_csv("data/raw/Sample - Superstore.csv", encoding="cp1252", dtype={"Postal Code": "str"})
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y").dt.normalize()
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y").dt.normalize()

federal_holidays = np.array([

    # 2014
    "2014-01-01", "2014-01-20", "2014-02-17",
    "2014-05-26", "2014-07-04", "2014-09-01",
    "2014-10-13", "2014-11-11", "2014-11-27", "2014-12-25",

    # 2015
    "2015-01-01", "2015-01-19", "2015-02-16",
    "2015-05-25", "2015-07-03", "2015-09-07",
    "2015-10-12", "2015-11-11", "2015-11-26", "2015-12-25",

    # 2016
    "2016-01-01", "2016-01-18", "2016-02-15",
    "2016-05-30", "2016-07-04", "2016-09-05",
    "2016-10-10", "2016-11-11", "2016-11-24", "2016-12-26",

    # 2017
    "2017-01-02", "2017-01-16", "2017-02-20",
    "2017-05-29", "2017-07-04", "2017-09-04",
    "2017-10-09", "2017-11-10", "2017-11-23", "2017-12-25",

    # Needed because the dataset includes shipments in early 2018
    "2018-01-01",

], dtype="datetime64[D]")
```
